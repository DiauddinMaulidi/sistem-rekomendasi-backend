```python
df = newData.rename(columns={
    'Soil_pH': 'pH_Tanah',
    'Soil_Moisture': 'Kelembaban',
    'Nitrogen_Level': 'Nitrogen',
    'Phosphorus_Level': 'Fosfor',
    'Potassium_Level': 'Kalium',
    'Temperature': 'Suhu_Tanah',
    'Crop_Type': 'Jenis_Tanaman',
    'Recommended_Fertilizer': 'Jenis_Pupuk'
})
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
y_label = le.fit_transform(df['Jenis_Pupuk'])
y_label
```

```python
print(le.classes_)
```

```python
X = df[['pH_Tanah', 'Kelembaban', 'Electrical_Conductivity', 'Nitrogen', 'Fosfor', 'Kalium', 'Suhu_Tanah', 'Jenis_Tanaman']]
y = y_label
```

```python
data_jenis_tanaman = pd.get_dummies(X, columns=['Jenis_Tanaman'])
data_jenis_tanaman
```

```python
X_train, X_test, y_train, y_test = train_test_split(data_jenis_tanaman, y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scale = scaler.fit_transform(X_train)
X_train_scale
```

```python
X_test_scale = scaler.transform(X_test)
X_test_scale
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
```

```python
model.fit(X_train_scale, y_train)
```

```python
y_pred = model.predict(X_test_scale)
y_pred
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
hasil = le.inverse_transform(y_pred)
hasil
```

```python
print(classification_report(y_test, y_pred))
```
